# Vehicle Sales - Exploratory Data Analysis

```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('../vehicles_us.csv')
df.head()
```

```python
df.info()
```

```python
fig = px.histogram(df, x='odometer', title='Odometer Distribution')
fig.show()
```

```python
fig = px.scatter(df, x='odometer', y='price', title='Price vs Odometer')
fig.show()
```
